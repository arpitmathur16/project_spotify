Top 10 artists consistently drive the highest average track popularity on the platform?


```python
top_artists = (
    df.groupby('artist_name')['track_popularity']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)
```

```python
# Spotify colors
bg_color = '#121212'
bar_color = '#1DB954'
text_color = '#FFFFFF'
grid_color = '#B3B3B3'

# Plot
plt.figure(figsize=(12, 6))
plt.gcf().patch.set_facecolor(bg_color)
plt.gca().set_facecolor(bg_color)

plt.barh(top_artists.index, top_artists.values, color=bar_color)

# Highest on top
plt.gca().invert_yaxis()

# Titles and labels
plt.title('Top 10 Artists by Average Track Popularity',
          fontsize=16, color=text_color, fontweight='bold')
plt.xlabel('Average Track Popularity', color=text_color)
plt.ylabel('Artist Name', color=text_color)

# Tick colors
plt.gca().tick_params(colors=text_color)

# Grid
plt.grid(axis='x', linestyle='--', alpha=0.3, color=grid_color)

# Value labels
for i, v in enumerate(top_artists.values):
    plt.gca().text(v + 0.5, i, f"{v:.2f}", va='center', color=text_color)

# Remove spines for clean UI look
for spine in plt.gca().spines.values():
    spine.set_visible(False)

plt.tight_layout()
plt.show()
```


What proportion of the platform's catalog consists of explicit vs non-explicit content?

```python
explicit_counts =  df['explicit'].value_counts()
```

```python
# Labels
labels = ['Non-Explicit', 'Explicit']

# Theme colors
colors = ['#1DB954', '#2A2A2A']
bg_color = '#000000'
text_color = '#FFFFFF'

# Plot
plt.figure(figsize=(6, 6))
plt.gcf().patch.set_facecolor(bg_color)
plt.gca().set_facecolor(bg_color)

wedges, texts, autotexts = plt.pie(
    explicit_counts,
    labels=labels,
    autopct='%1.1f%%',
    startangle=90,
    colors=colors,
    wedgeprops={'edgecolor': 'white', 'linewidth': 1.5},
    textprops={'color': text_color, 'fontsize': 11}
)

# Title
plt.title('Explicit vs Non-Explicit Tracks',
          fontsize=14, color=text_color, fontweight='bold')

# Style percentage text
for autotext in autotexts:
    autotext.set_color('white')
    autotext.set_fontsize(11)

plt.tight_layout()
plt.show()
```

Track duration vary across album types — and are there significant outliers?

```python
palette = {'album': '#1DB954', 'single': '#F6C90E', 'compilation': '#1EAAFF'}
types   = ['album', 'single', 'compilation']

# Figure setup
plt.figure(figsize=(14, 8))
plt.gcf().patch.set_facecolor('#191414')
plt.gca().set_facecolor('#121212')

# One box per album type
for i, t in enumerate(types, 1):
    grp = df[df['album_type'] == t]['track_duration_min'].dropna().values
    c   = palette[t]
    med, q1, q3 = np.median(grp), *np.percentile(grp, [25, 75])

    # Draw boxplot
    plt.boxplot(grp, positions=[i], widths=0.48, patch_artist=True,
                boxprops     = dict(facecolor=c + '33', color=c),
                medianprops  = dict(color='white'),
                whiskerprops = dict(color=c),
                capprops     = dict(color=c))

    # Annotate median, Q1, Q3
    plt.gca().text(i + 0.3, med, f'Med: {med:.2f}', color='white', fontsize=8,   va='center')
    plt.gca().text(i + 0.3, q3,  f'Q3 : {q3:.2f}',  color=c,      fontsize=7.5, va='bottom')
    plt.gca().text(i + 0.3, q1,  f'Q1 : {q1:.2f}',  color=c,      fontsize=7.5, va='top')

# Labels and styling
plt.xticks([1, 2, 3], ['Album', 'Single', 'Compilation'], color='white')
plt.title('Track Duration by Album Type', color='white', fontsize=14, fontweight='bold')
plt.gca().tick_params(colors='#B3B3B3')
plt.gca().yaxis.grid(True, color='#282828', linestyle='--')

plt.tight_layout()
plt.show()
```

Artists command the largest listener base, and dominant in their market share

```python
top5 = (df.drop_duplicates('artist_name')
          .nlargest(5, 'artist_followers')
          .set_index('artist_name')['artist_followers'])

ax = top5.plot(
    kind    = 'bar',
    color   = ['#1DB954', '#1EAAFF', '#F6C90E', '#E8834C', '#B34FFF'],
    figsize = (10, 6),
    title   = 'Top 5 Artists by Follower Count'
)

# Background
ax.figure.patch.set_facecolor('#191414')
ax.set_facecolor('#121212')

# Labels & Title
ax.set_xlabel('Artist',    color='white', fontsize=11)
ax.set_ylabel('Followers', color='white', fontsize=11)
ax.title.set_color('white')

# Tick Colors
ax.tick_params(colors='white')
for t in ax.get_xticklabels():
    t.set_color('white')

# Grid
ax.yaxis.grid(True, color='#282828', linestyle='--')
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```

Music production volume trended over the years — and peak it

```python
df['year'] = pd.to_datetime(df['album_release_date'], errors='coerce').dt.year
yearly = df.groupby('year')['track_name'].count()

ax = yearly.plot(kind='line', figsize=(9, 4), color='#1DB954', linewidth=2, marker='o', markersize=4)

ax.set_title('Music Production Volume Over the Years', fontsize=12)
ax.set_xlabel('Release Year')
ax.set_ylabel('Number of Tracks')
ax.spines[['top', 'right']].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
ax = df.groupby(['album_type', 'explicit'])['track_popularity'].mean().unstack().plot(
    kind='barh', figsize=(8, 4), color=['#4CAF50', '#E53935'], alpha=0.85
)

for bar in ax.patches:
    ax.text(bar.get_width() + 0.2, bar.get_y() + bar.get_height() / 2,
            f'{bar.get_width():.1f}', va='center', fontsize=9)

ax.set_title('Avg Track Popularity by Album Type & Explicit', fontsize=12)
ax.set_xlabel('Average Popularity')
ax.legend(['Clean', 'Explicit'])
ax.spines[['top', 'right']].set_visible(False)

plt.tight_layout()
plt.show()
```

Albums delivers the consistent popularity across all their tracks

```python
top10 = (df.groupby('album_name')['track_popularity']
           .mean()
           .sort_values(ascending=False)
           .head(10))

# Figure setup
plt.figure(figsize=(13, 7))
plt.gcf().patch.set_facecolor('#191414')
plt.gca().set_facecolor('#121212')

# Bar chart
plt.barh(top10.index, top10.values, color='#1DB954', edgecolor='white')

# Titles and labels
plt.title('Top 10 Most Consistently Popular Albums', color='white', fontsize=14, fontweight='bold')
plt.xlabel('Avg Popularity', color='#B3B3B3')
plt.xticks(color='white')
plt.yticks(color='white')

# Grid
plt.gca().xaxis.grid(True, color='#282828', linestyle='--')

plt.tight_layout()
plt.show()
```

```python
cols = ['track_popularity', 'artist_popularity', 'artist_followers',
        'track_duration_min', 'album_total_tracks']
corr = df[cols].corr().round(2)

# Figure setup
plt.figure(figsize=(16, 8))
plt.gcf().patch.set_facecolor('#191414')
plt.gca().set_facecolor('#121212')

# Heatmap
sns.heatmap(corr, annot=True, fmt='.2f', cmap='viridis',
            linewidths=0.5, linecolor='#191414',
            annot_kws={'size': 11, 'weight': 'bold', 'color': 'white'},
            square=True, vmin=-1, vmax=1, ax=plt.gca())

# Titles and ticks
plt.title('What Drives Track Popularity?', color='white', fontsize=14, fontweight='bold', pad=15)
plt.gca().tick_params(colors='white', length=0)
plt.setp(plt.gca().get_xticklabels() + plt.gca().get_yticklabels(), color='white')

plt.tight_layout()
plt.show()
```
